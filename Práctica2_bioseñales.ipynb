# IMPULSO UNITARIO

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#Ejemplo
#n = np.arange(-2, 4) #Siempre se le suma uno al intervalo máximo, es decir, aqui cuenta hasta 2
#x = np.array([5, 6, 7, 0, 4, 3,8]) , secuencia definida solo en instantes de tiempo

def impulso_unitario(n0,n1,n2):# Señal impulso: δ[n - n0]

#n0 momento en el que el impulso debe tener un valor de 1
#n1 valor inicio del rango
#n2 valor final del rango
  n=np.arange(n1,n2+1) # Eje discreto
  x= (n == n0).astype(int) #Se hace una comparacion , para mirar si mi n0, es igual a 0 , pone un true(1)-->impulso unitario
#n: [5, 6, 7, 0, 4, 3,8]
#x:  0  0  0 1 0 0 0
  return x, n

x,n = impulso_unitario(0,-5,5) #valores para poder ver la gráfica, porque es una función . La podemos llamar en cualquier momento

plt.stem(n,x)#Graficar señales discretas
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("x") #Nombre al eje vertical
plt.title("Impulso unitario")#nombre a la función
plt.show()#Mostrar gráfica


```

# ESCALÓN UNITARIO

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def escalon_unitario(n0,n1,n2):

  n=np.arange(n1,n2+1)    # Se hace una camparacion , para mieae si mi n2, es igual a 0 , pone un true(1)
  x= ( n>= n0 ).astype(int) #Mayor o igual a 0, vale 1. Antes vale 0
  return x, n

x,n=escalon_unitario(0,-5,5) #valores gráfica
plt.stem(n,x);#Gráfica discreta


```

# RAMPA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def rampa(n0,n1,n2, m=1): # Hay que hacer que tenga pendiente, porque los valoeres aumentan linealmente desde 0

  n=np.arange(n1,n2+1)    # Se hace una camparacion , para mieae si mi n2, es igual a 0 , pone un true(1)
  x= m*( n>= n0 )*(n-n0).astype(int) # n =t, n0=0 , y si es mayor a cero tiene que seguir subiendo
  return x, n

x,n=rampa(0,10,50)
plt.stem(n,x); #; es para quitar el texto que suelta con stem
```

# EXPONENCIALES COMPLEJAS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def exponencial_compleja(sigma, omega0, n0, n1):
  # Sigma= Atenuación(decae) si es negativa , o una amplificación(crece) si es positivo, y si es 0 la amplitud es constante
  # w(omega)= Frecuencia

  n = np.arange(n0, n1 + 1) #Eje discreto

  e = (sigma + 1j * omega0) * n # Expresión de la exponencial compleja
  # Parte real y parte imaginaria

  # Fórmula de la exponencial compleja
  x = np.exp(e)

  return x, n

x,n=exponencial_compleja(0, 0.3*np.pi,0,50)#Valores para aplicar
plt.stem(n,x);


```

# SINUSOIDES





```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def sinusoide(phi0, omega0, n1, n2):
# Parametros
# Omega: Angulo
# Phi: Desplazamiento

  n = np.arange(n1, n2 + 1) # Puntos donde se evaluará la señal

  # Fórmula Función del seno
  x = np.sin((omega0 * n + phi0))


  return x, n

x,n = sinusoide(0, 0.2*np.pi, 0, 50) # Valores para ver lañ función
plt.stem(n,x)# Gráfica discreta
plt.xlabel("n")
plt.ylabel("x[n]")
plt.title("Sinusoide")
plt.show()
```
