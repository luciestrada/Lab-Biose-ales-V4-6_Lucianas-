# **CONSULTA**
### **Funciones de numpy:**

*Numpy.ones():*  

Esta función permite crear una matriz de unos con la forma y tipo determinados

**numpy.ones(forma, tipo de dato = Ninguno, orden = 'C')**

*   **forma:** es el numero de columnas y filas de la matriz (2) , (2,2) , (2,3)
*   **tipo de dato:** por predeterminado se encuentra en flotante, entonces para cambiarlo se agrega dtype=int


---


*Numpy.round():*

Se encarga de redondear uniformemente al número dado de decimales

**numpy.round ( a , decimales = 0 , salida = Ninguno )**


*   **a:** Es el valor a redondear (datos de entrada)
*   **decimales:** Es la cantidad de decimales a redondear, el valor predeterminado en 0
*  **salida:** Matriz opcional donde se guardará el resultado. Debe tener la misma forma (dimensiones) que la salida esperada.


---

*Numpy.ceils()*

Esta función permite devolver el entero más cercano que sea **mayor o igual** al número original.

**numpy.ceil(x)**


*  **x:** Números a redondear

---
*Numply.floor()*

Esta función permite devolver el entero más cercano que sea **menor o igual** al número original.

**numpy.floor(x)**


*  **x:** Números a redondear



```python
#Ejemplo de Numpy.ones
A= np.ones((5,3), dtype=int)
B= np.ones((2), dtype=int)

# Ejemplo de Numpy.round
C= np.round(7646546.89779809, 3)

#Ejemplo de Numpy.ceils
D = np.array([1.2, 3.7, 4.1])

#Ejemplo de Numpy.floor
E = np.array([-1.7, -1.5, -0.2, 0.2, 1.5, 1.7, 2.0])

print("Matriz de 1 con 5 filas y 3 columnas y números enteros", A)
print("Matriz de dos columnas", B )
print("Redondeado a 3 decimales", C)
print("Números redondeados a enteros mayores" , np.ceil(D))

print("Números redondeados a enteros menores" , np.floor(E))
```

```python
#FUNCION 1
# Crear el intervalo de n desde 0 hasta 100
n = np.arange(0, 101)

# Calcular la función
def y(n):
  return np.sin(np.pi * 0.12 * n)
yn=y(n)

plt.stem(n, yn) #Se usa para señales discretas como y[n] y coloca puntos en cada valor
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("y(n)") #Nombre al eje vertical
plt.title("y[n] = sin(π * 0.12n)")#nombre a la función
plt.show()#Mostrar gráfica
```

```python
 #FUNCIÓN 2

n = np.arange(0, 101) #Este intervalo es de 0 a 100

# Calcular la función
def y2(n):
  return np.cos( 2 * np.pi * 0.03 * n )
y2n= y2(n)

plt.stem(n, y2n) #Se usa para señales discretas y coloca puntos en cada valor
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("y(n)") #Nombre al eje vertical
plt.title("y2[n] = cos(2π ∗ 0.03n)")#nombre a la función
plt.show()#Mostrar gráfica
```

```python
#SUMA DE LAS FUNCIONES
s=yn+y2n

#MULTIPLICAIÓN DE LAS FUNCIONES
t=yn*y2n

plt.stem(n, s) #Se usa para señales discretas y coloca puntos en cada valor
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("s(n)") #Nombre al eje vertical
plt.title("s(n)= suma de funciones")#nombre a la función
plt.show()#Mostrar gráfica

plt.stem(n, t) #Se usa para señales discretas y coloca puntos en cada valor
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("t(n)") #Nombre al eje vertical
plt.title("t(n)= multiplicación de funciones")#nombre a la función
plt.show()#Mostrar gráfica
```

```python
#Gráfica de dos señales en una misma figura

plt.stem(n, yn, linefmt= 'g' , markerfmt= 'g', basefmt='y' )
plt.stem(n, y2n, linefmt= 'm' , markerfmt= 'm', basefmt='y' )
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("y(n) y y2(n)") #Nombre al eje vertical
plt.title("Funciones y[n] y y2[n]")#nombre a la función
plt.show()#Mostrar gráfica

plt.stem(n, s, linefmt= 'b' , markerfmt= 'b', basefmt='k' ) # si no fuera una señal discreta se utilizaria plt.plot
plt.stem(n, t, linefmt= 'c' , markerfmt= 'c', basefmt='k' )
plt.xlabel("n") # Nombre al eje horizontal
plt.ylabel("s(n) y t(n)") #Nombre al eje vertical
plt.title("Funciones s[n] y t[n]")#nombre a la función
plt.show()#Mostrar gráfica
```

**1)** Escribir una función que reciba un diccionario con las notas de los alumnos de un curso y devuelva una serie con la nota mínima, la máxima, media y la desviación típica.


**2)** Cálculo del Índice de Masa Corporal (IMC) con Pandas.

```python
# Diccionario con las notas
notas = {
    'Sara': 4.7,
    'Kevin': 5.0,
    'Ángel': 3.0,
    'Matías': 1.4,
    'Josefina': 3.7,
    'Carlos': 2.9,
    'Juana': 3.3
}

# Función que calcula estadísticas
def estadisticas_notas(diccionario):

    # Convertir diccionario en Serie
    serie = pd.Series(diccionario)

    # Calcular estadísticas
    resultado = pd.Series({
        "Minima": serie.min(),
        "Maxima": serie.max(),
        "Media": serie.mean(),
        "Desviacion_tipica": serie.std()
    })

    return resultado.round(3)

# Llamar la función e imprimir resultado
print(estadisticas_notas(notas))
```
